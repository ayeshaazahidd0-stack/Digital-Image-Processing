```python
!pip install -q opencv-python
import cv2, numpy as np
from google.colab.patches import cv2_imshow
from google.colab import files
```

```python
uploaded = files.upload()
fname = list(uploaded.keys())[0]
img = cv2.imread(fname)
print("Loaded:", fname)
```

```python
print("Shape (H, W, C):",img.shape)
```

```python
cv2_imshow(img)
```

```python
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
cv2.imwrite("Practice1_gray.jpg",gray)
print("Saved Practice1_gray.jpg")
```

```python
avg = float(np.mean(gray))
print("Average brightness:",round(avg,2))
```

```python
_,bw = cv2.threshold(gray,128,255,cv2.THRESH_BINARY)
cv2_imshow(bw)
```

```python
h,w = gray.shape
small = cv2.resize(gray,(w//2,h//2))
cv2_imshow(small)
```

```python
flipped = cv2.flip(img, 1)
cv2_imshow(flipped)
```

```python
rect_img = img.copy()
cv2.rectangle(rect_img,(50,50),(200,200),(0,255,0),3)
cv2_imshow(rect_img)
```
